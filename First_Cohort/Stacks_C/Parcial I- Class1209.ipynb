```
Procedimiento BubbleSort(arr, n)

        swapped ← falso
        Para i desde 0 hasta n - 2 hacer

                temp ← arr[i]
                arr[i] ← arr[i + 1]
                arr[i + 1] ← temp
                swapped ← verdadero

    Hasta que swapped = falso
```

**a.** Evaluar si el código presentado funciona correctamente tal como fue presentado o si es necesario realizar alguna corrección a la función. En caso de que sea necesario realizar alguna modificación, presente la versión corregida de la función.

````c++
void BubbleSort(int arr[], int n) {
    bool swapped;
    do {
        swapped = false;

        for (int i = 0; i <= n - 2; i++) {
            if (arr[i] < arr[i + 1]) {
                int temp = arr[i];
                arr[i] = arr[i + 1];
                arr[i + 1] = temp;
                swapped = true;
            }
        }

    } while (swapped == true);
}
````

````c++
typedef struct SLLNode {
    int data;
    struct SLLNode* next;
} SLLNode;

// Create a new node
SLLNode* createSLLNode(int data) {
    SLLNode* newNode = (SLLNode*)malloc(sizeof(SLLNode));
    newNode->data = data;
    newNode->next = NULL;
    return newNode;
}

// Create a circular singly linked list with n elements
SLLNode* createCircularSLL(int n) {
    if (n <= 0) return NULL;

    SLLNode* head = createSLLNode(1);
    SLLNode* temp = head;

    for (int i = 2; i <= n; i++) {
        temp->next = createSLLNode(i);
        temp = temp->next;
    }

    temp->next = head; // last node points back to the head
    return head;
}

// Print the circular list (one full cycle of n elements)
void printCircularSLL(SLLNode* head, int n) {
    if (head == NULL) return;

    SLLNode* temp = head;
    for (int i = 0; i <= n; i++) {
        printf("%d ", temp->data);
        temp = temp->next;
    }
    printf("\n");
}
````

### El algoritmo de ordenamiento SelectionSort tiene complejidad en el peor caso O(n²). En promedio…
- a) Es más eficiente que Quicksort.
- b) Tiene el mismo rendimiento que Quicksort.

````c++
// Swap two integers
void swap(int *a, int *b) {
    int temp = *a;
    *a = *b;
    *b = temp;
}

// Print array
void printArray(int arr[], int n) {
    for (int i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }
    printf("\n");
}

// ==========================
// Selection Sort (O(n^2))
// ==========================
void selectionSort(int arr[], int n) {
    printf("\n Selection Sort \n");
    for (int i = 0; i < n - 1; i++) {
        int minIndex = i; // Set the first unsorted element as the minimum
        // Search for the minimum in the remaining unsorted elements
        for (int j = i + 1; j < n; j++) {
            if (arr[j] < arr[minIndex]) {
                minIndex = j; // Found new minimum
            }
        }
        // Swap minimum element with the first unsorted element
        if (minIndex != i) {
            swap(&arr[i], &arr[minIndex]);
        }

        printf("Pass %d: ", i + 1);
        printArray(arr, n);
    }
}

// ==========================
// Quicksort (O(n log n))
// ==========================
// Partition function (based on VisuAlgo)
int partition(int arr[], int left, int right, int n, int depth) {
    int pivotIndex = left;
    int pivotValue = arr[pivotIndex];
    int storeIndex = pivotIndex + 1;

    for (int i = pivotIndex + 1; i <= right; i++) {
        // If element < pivot, or equal but "50% lucky"
        if ((arr[i] < pivotValue) ||
            (arr[i] == pivotValue && rand() % 2 == 0)) {
            swap(&arr[i], &arr[storeIndex]);
            storeIndex++;
        }
    }
    // Place pivot in its correct position
    swap(&arr[pivotIndex], &arr[storeIndex - 1]);

    // Print trace with indentation = depth
    for (int i = 0; i < depth; i++) printf("  ");
    printf("Partition (pivot=%d): ", pivotValue);
    printArray(arr, n);
    // return new pivot index
    return storeIndex - 1;
}

// Internal recursive quicksort (with depth and tracing)
void quicksort_internal(int arr[], int left, int right, int n, int depth) {
    if (left < right) {
        printf("\n Quick Sort \n");
        int pivotNew = partition(arr, left, right, n, depth);
        quicksort_internal(arr, left, pivotNew - 1, n, depth + 1);
        quicksort_internal(arr, pivotNew + 1, right, n, depth + 1);
    }
}

// Public quicksort function (clean interface)
void quicksort(int arr[], int left, int right) {
    int n = right + 1; // array size
    quicksort_internal(arr, left, right, n, 0);
}
````

Presente una alternativa de solución para el problema de determinar si una expresión que contiene paréntesis (), corchetes [] y/o llaves {} está bien balanceada; esto es, que todo símbolo de apertura tenga posteriormente un símbolo de cierre en el orden correcto, empleando la estructura de datos de su elección. Esta alternativa puede ser presentada por medio de una descripción detallada de los pasos que debe seguir el algoritmo, pseudocódigo o utilizando el lenguaje de programación C.

### Pseudocódigo

```
function isBalanced(expression):
    create empty stack

    for each character c in expression:
        if c is '(', '[', or '{':
            push c onto stack
        else if c is ')', ']', or '}':
            if stack is empty:
                return "not balanced"
            top = pop from stack
            if c does not match top:
                return "not balanced"

    if stack is empty:
        return "balanced"
    else:
        return "not balanced"
```

````c++
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <stdbool.h>

// ==========================
// Stack implementation using Linked List
// ==========================

// Node structure (each node stores a character)
typedef struct NodeChar {
    char data;               // Symbol ( '(', '[', '{' )
    struct NodeChar *next;   // Pointer to the next node
} NodeChar;

// Stack structure (just holds a pointer to the top node)
typedef struct {
    NodeChar *top;           // Top of the stack
} Stack;

// Create an empty stack
Stack* createStack() {
    Stack *stack = (Stack*)malloc(sizeof(Stack));
    stack->top = NULL;
    return stack;
}

// Check if the stack is empty
bool isEmpty(Stack *stack) {
    return stack->top == NULL;
}

// Push a character onto the stack
void push(Stack *stack, char c) {
    NodeChar *newNode = (NodeChar*)malloc(sizeof(NodeChar));
    newNode->data = c;
    newNode->next = stack->top;  // Link new node to previous top
    stack->top = newNode;        // Update top
}

// Pop the top character from the stack
char pop(Stack *stack) {
    if (isEmpty(stack)) return '\0'; // If empty, return null char
    NodeChar *temp = stack->top;
    char popped = temp->data;
    stack->top = temp->next; // Move top down
    free(temp);              // Free memory
    return popped;
}

// Peek at the top character without removing it
char peek(Stack *stack) {
    if (isEmpty(stack)) return '\0';
    return stack->top->data;
}

// Free the entire stack (all nodes)
void freeStack(Stack *stack) {
    while (!isEmpty(stack)) {
        pop(stack);
    }
    free(stack);
}

// ==========================
// Helper function
// ==========================

// Check if opening and closing symbols match
bool isMatchingPair(char open, char close) {
    return (open == '(' && close == ')') ||
           (open == '[' && close == ']') ||
           (open == '{' && close == '}');
}

// ==========================
// Core algorithm
// ==========================

// Function to check if an expression is balanced
bool isBalanced(const char *expr) {
    Stack *stack = createStack();  // create an empty stack

    // Traverse the expression character by character
    for (int i = 0; expr[i] != '\0'; i++) {
        char c = expr[i];

        // If it's an opening symbol, push onto stack
        if (c == '(' || c == '[' || c == '{') {
            push(stack, c);
        } 
        // If it's a closing symbol, check for matching
        else if (c == ')' || c == ']' || c == '}') {
            if (isEmpty(stack)) {
                freeStack(stack);
                return false;  // closing without opening
            }
            char top = pop(stack);
            if (!isMatchingPair(top, c)) {
                freeStack(stack);
                return false;  // mismatched pair
            }
        }
    }

    // Expression is balanced if stack is empty at the end
    bool balanced = isEmpty(stack);
    freeStack(stack);
    return balanced;
}
````
