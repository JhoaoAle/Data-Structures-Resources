```c++
#include <stdio.h>
#include <stdlib.h>
#include <stdbool.h>
```

```c++
typedef struct Node {
    int data;
    struct Node* next;
} Node;
```

```c++
Node* create_node(int value){
    Node* new_Node = (Node *)malloc(sizeof(Node));
    new_Node->data = value;
    new_Node->next = NULL;
    return new_Node;
}
```

```c++
void print_list(Node* head){
    Node *temp = head;
    while (temp != NULL){
        printf("%d -> ", temp->data);
        temp = temp->next;
    }
    printf("NULL \n");
}
```

```c++
Node* insert_at_head(Node* head, int value){
    Node* new_Node = create_node(value);
    new_Node->next = head;
    return new_Node;
}
```

```c++
Node* insert_at_tail(Node* head, int value){
    Node* new_Node = create_node(value);
    Node *temp = head;
    while (temp->next != NULL){
        temp = temp->next;
    }
    temp->next = new_Node;
    return head;
}
```

```c++
void insert_at_head_void(Node** head, int value) {
    Node* new_Node = create_node(value);
    new_Node->next = *head;
    *head = new_Node;
}
```

```c++
void insert_at_tail_void(Node** head, int value) {
    Node* new_Node = create_node(value);

    // If list is empty
    if (*head == NULL) {
        *head = new_Node;
        return;
    }

    Node* temp = *head;
    while (temp->next != NULL) {
        temp = temp->next;
    }

    temp->next = new_Node;
}
```

```c++
Node *head = create_node(2);
head->next = create_node(3);
head->next->next = create_node(4);
head->next->next->next = create_node(5);


int x = 1;
head = insert_at_head(head, x);
head = insert_at_tail(head, x);
print_list(head);
```

```c++
int y = 10;
insert_at_head_void(&head, y);
insert_at_tail_void(&head, y);

print_list(head);
```

```c++
bool search_key(struct Node *head, int key)
{
    struct Node *curr = head;

    while (curr != NULL)
    {
        if (curr->data == key)
            return true;
        curr = curr->next;
    }

    return false;
}
```

```c++
Node* delete_head(struct Node *head)
{
    if (head == NULL)
        return NULL;

    struct Node *temp = head;
    head = temp->next;
    free(temp); // Free memory of old head

    return head;
}
```

```c++
head = delete_head(head);

print_list(head);
```

```c++
Node* delete_tail(struct Node *head)
{
    if (head == NULL)
        return NULL;

    if (head->next == NULL)
    {
        free(head);
        return NULL;
    }

    struct Node *secondLast = head;
    while (secondLast->next->next != NULL)
    {
        secondLast = secondLast->next;
    }

    free(secondLast->next);
    secondLast->next = NULL;
    return head;
}
```

```c++
head = delete_tail(head);

print_list(head);
```

```c++
Node* delete_at_position(Node* head, int position)
{
    if (head == NULL)
        return NULL;

    if (position == 0)
    {
        Node* temp = head;
        head = head->next;
        free(temp);
        return head;
    }

    Node* curr = head;

    // Move to node before target
    for (int i = 0; i < position - 1 && curr->next != NULL; i++)
    {
        curr = curr->next;
    }

    // If position is out of bounds
    if (curr->next == NULL)
        return head;

    Node* temp = curr->next;
    curr->next = temp->next;
    free(temp);

    return head;
}
```

```c++
head = delete_at_position(head, 1);

print_list(head);
```

```c++
Node* delete_by_value(Node* head, int value)
{
    if (head == NULL)
        return NULL;

    // If head contains the value
    if (head->data == value)
    {
        Node* temp = head;
        head = head->next;
        free(temp);
        return head;
    }

    Node* curr = head;

    while (curr->next != NULL && curr->next->data != value)
    {
        curr = curr->next;
    }

    // Value not found
    if (curr->next == NULL)
        return head;

    Node* temp = curr->next;
    curr->next = temp->next;
    free(temp);

    return head;
}
```

```c++
head = delete_by_value(head, 1);

print_list(head);
```
