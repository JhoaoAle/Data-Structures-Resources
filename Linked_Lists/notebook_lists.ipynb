```c++
#include <stdio.h>
#include <stdlib.h>
#include <stdbool.h>
```

```c++
typedef struct Node {
    int data;
    struct Node* next;
} Node;
```

```c++
Node* create_node(int value){
    Node* new_Node = (Node *)malloc(sizeof(Node));
    new_Node->data = value;
    new_Node->next = NULL;
    return new_Node;
}
```

```c++
void print_list(Node* head){
    Node *temp = head;
    while (temp != NULL){
        printf("%d -> ", temp->data);
        temp = temp->next;
    }
    printf("NULL \n");
}
```

```c++
Node* insert_at_head(Node* head, int value){
    Node* new_Node = create_node(value);
    new_Node->next = head;
    return new_Node;
}
```

```c++
Node* insert_at_tail(Node* head, int value){
    Node* new_Node = create_node(value);
    Node *temp = head;
    while (temp->next != NULL){
        temp = temp->next;
    }
    temp->next = new_Node;
    return head;
}
```

```c++
void insert_at_head_void(Node** head, int value) {
    Node* new_Node = create_node(value);
    new_Node->next = *head;
    *head = new_Node;
}
```

```c++
void insert_at_tail_void(Node** head, int value) {
    Node* new_Node = create_node(value);

    // If list is empty
    if (*head == NULL) {
        *head = new_Node;
        return;
    }

    Node* temp = *head;
    while (temp->next != NULL) {
        temp = temp->next;
    }

    temp->next = new_Node;
}
```

```c++
bool search_key(struct Node *head, int key)
{
    struct Node *curr = head;

    while (curr != NULL)
    {
        if (curr->data == key)
            return true;
        curr = curr->next;
    }

    return false;
}
```

```c++
Node* delete_head(struct Node *head)
{
    if (head == NULL)
        return NULL;

    struct Node *temp = head;
    head = temp->next;
    free(temp); // Free memory of old head

    return head;
}
```

```c++
head = delete_head(head);

print_list(head);
```

```c++
Node* delete_tail(struct Node *head)
{
    if (head == NULL)
        return NULL;

    if (head->next == NULL)
    {
        free(head);
        return NULL;
    }

    struct Node *secondLast = head;
    while (secondLast->next->next != NULL)
    {
        secondLast = secondLast->next;
    }

    free(secondLast->next);
    secondLast->next = NULL;
    return head;
}
```

```c++
head = delete_tail(head);

print_list(head);
```

```c++
Node* delete_at_position(Node* head, int position)
{
    if (head == NULL)
        return NULL;

    if (position == 0)
    {
        Node* temp = head;
        head = head->next;
        free(temp);
        return head;
    }

    Node* curr = head;

    // Move to node before target
    for (int i = 0; i < position - 1 && curr->next != NULL; i++)
    {
        curr = curr->next;
    }

    // If position is out of bounds
    if (curr->next == NULL)
        return head;

    Node* temp = curr->next;
    curr->next = temp->next;
    free(temp);

    return head;
}
```

```c++
head = delete_at_position(head, 1);

print_list(head);
```

```c++
Node* delete_by_value(Node* head, int value)
{
    if (head == NULL)
        return NULL;

    // If head contains the value
    if (head->data == value)
    {
        Node* temp = head;
        head = head->next;
        free(temp);
        return head;
    }

    Node* curr = head;

    while (curr->next != NULL && curr->next->data != value)
    {
        curr = curr->next;
    }

    // Value not found
    if (curr->next == NULL)
        return head;

    Node* temp = curr->next;
    curr->next = temp->next;
    free(temp);

    return head;
}
```

```c++
Node* insert_at_n(Node* head, int value, int n){
    if (n==0) return insert_at_head(head, value);
    if (head == NULL){
        printf("Out of bounds. Failed\n");
        return head;
    }
    Node* new_Node = create_node(value);
    Node* curr = head;
    int tracker = 0;
    while (tracker < n - 1){
        if(curr->next == NULL){
            printf("Out of bounds. Failed");
            free(new_Node);
            return head;
        }
        curr = curr->next;
        tracker += 1;
    }
    
    new_Node->next = curr->next;
    curr->next = new_Node;
    
    return head;
}
```

```c++
typedef struct LinkedList {
    Node* head;
    Node* tail;
} LinkedList;
```

```c++
LinkedList* create_list(){
    LinkedList* list = (LinkedList*)malloc(sizeof(LinkedList));
    list->head = list->tail = NULL;
    return list;
}
```

```c++
void insert_at_tail(LinkedList* list, int value){
    Node* new_node = create_node(value);
    if (list->tail != NULL){
        list->tail->next = new_node;
        list->tail = new_node;
    } else {
        list->head = new_node;
        list->tail = new_node;
    }
}
```

```c++
void delete_head(LinkedList* list)
{
    if (list->head == NULL)
        return;

    Node* temp = list->head;
    list->head = temp->next;

    // If list becomes empty
    if (list->head == NULL) list->tail = NULL;

    free(temp);
}
```

```c++
void delete_tail(LinkedList* list)
{
    if (list->head == NULL) return;

    // One node
    if (list->head == list->tail) {
        free(list->head);
        list->head = NULL;
        list->tail = NULL;
        return;
    }

    Node* curr = list->head;

    while (curr->next != list->tail) curr = curr->next;

    free(list->tail);
    curr->next = NULL;
    list->tail = curr;
}
```

```c++
void delete_at_n(LinkedList* list, int position)
{
    if (list->head == NULL) return;

    // Delete head
    if (position == 0)
    {
        Node* temp = list->head;
        list->head = temp->next;

        if (list->head == NULL) list->tail = NULL;

        free(temp);
        return;
    }

    Node* curr = list->head;
    for (int i = 0; i < position - 1 && curr->next != NULL; i++) curr = curr->next;

    if (curr->next == NULL) return;

    Node* temp = curr->next;
    curr->next = temp->next;

    // If deleted tail
    if (temp == list->tail) list->tail = curr;
    free(temp);
}
```

```c++
void delete_by_value(LinkedList* list, int value)
{
    if (list->head == NULL) return;

    // If head contains the value
    if (list->head->data == value)
    {
        Node* temp = list->head;
        list->head = list->head->next;

        if (list->head == NULL) list->tail = NULL;
        free(temp);
        return;
    }

    Node* curr = list->head;

    while (curr->next != NULL && curr->next->data != value) curr = curr->next;

    // Value not found
    if (curr->next == NULL) return;

    Node* temp = curr->next;
    curr->next = temp->next;
    if (temp == list->tail) list->tail = curr;
    free(temp);

    return;
}
```
