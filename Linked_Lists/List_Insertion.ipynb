## Defining the structure of a Node

A node is defined as a pair of elements:

- The value inside the Node (in the case a parameter named data that ought to contain an integer)
- A pointer to the next node in the Linked List, since it has to point to a Node, it must be defined as a pointer to a type Node, as explored during the class on how pointers work.  As a reminder, pointers are **strongly typed** so the types of the pointer and the pointed memory must match

```c++
typedef struct Node {
    int data;
    struct Node* next;
} Node;
```

## Creation of a Node

In order to create a node, we use the following function, considering:

- The function will return a pointer to a Node.

- It takes an integer value as input.

- `malloc(sizeof(Node))` allocates enough memory to store a Node. No more, no less, however malloc by itself **does not know** to what purpose it is being used, it just sees it needs a space in memory. As such, the function as is returns a generic `void *` pointer. Since we need to be specifically a pointer for a Node, it has to be casted to be instead a pointer to the memory space of a node.

  In the current implementation this is achieved by the `(Node *)` prefix.

  Check if this is neccesary by removing the casting and see if C is able to automatically modify this allocated memory pointer type without it.

  Now that a have a memory space for a node is being returned by this operation, we can assign it to the `new_Node` pointer.

- The provided value is stored inside the Node’s `data` field.

- By default, this node is not linked to any other Node, so it starts pointing to `NULL`

```c++
Node* create_node(int value){
    Node* new_Node = (Node *)malloc(sizeof(Node));
    new_Node->data = value;
    new_Node->next = NULL;
    return new_Node;
}
```

## Printing a Linked List

Starting at the head of the linked list, traverse it via the `next` pointers of each Node, printing the value inside.

It stops when it hits a `NULL`.

What would happen if it stopped at the value inmediately before that? How would that be implemented?

```c++
void print_list(Node* head){
    Node *temp = head;
    while (temp != NULL){
        printf("%d -> ", temp->data);
        temp = temp->next;
    }
    printf("NULL \n");
}
```

## Head Insertion

Inserting at the head of a list means changing the head for whatever Node we are inserting and then linking the previous head to the new one.
Since this returns the `new_Node`, it would be returning the head and it can be assigned as such when the function is called.

```c++
Node* insert_at_head(Node* head, int value){
    Node* new_Node = create_node(value);
    new_Node->next = head;
    return new_Node;
}
```

## Tail Insertion

Inserting at the tail of a list means changing the last element of the Linked List to point to my `new_Node` instead of pointing at `NULL`.

Since this changes the tail, `new_Node` would make no sense as the returned value. Instead, as the implementation interacts with the head in general, we retun the same `head` we received.

```c++
Node* insert_at_tail(Node* head, int value){
    Node* new_Node = create_node(value);
    Node *temp = head;
    while (temp->next != NULL){
        temp = temp->next;
    }
    temp->next = new_Node;
    return head;
}
```

## Proof of Concept

```c++
Node *head = create_node(2);
head->next = create_node(3);
head->next->next = create_node(4);
head->next->next->next = create_node(5);


int x = 1;
head = insert_at_head(head, x);
head = insert_at_tail(head, x);
print_list(head);
```
