```c++
#include <stdlib.h>
#include <stdio.h>
```

```c++
// Definition of the Heap structure (Max-Heap)
typedef struct {
    int* array;      // The Heap array (starts at index 1, index 0 is unused)
    int size;        // Current number of elements
    int capacity;    // Maximum capacity
} Heap;
```

```c++
// Heapify function: maintains the max-Heap property from a given index downwards
void maxHeapify(Heap* h, int index)
{
    int left = 2 * index;
    int right = 2 * index + 1;
    int largest = index;

    // Compare with left child
    if (left <= h->size && h->array[left] > h->array[largest])
        largest = left;
    // Compare with right child
    if (right <= h->size && h->array[right] > h->array[largest])
        largest = right;

    // If one of the children is larger, swap and recurse
    if (largest != index) {
        int temp = h->array[largest];
        h->array[largest] = h->array[index];
        h->array[index] = temp;

        maxHeapify(h, largest);
    }
}
```

```c++
// Function to create a Heap from an array
Heap* createHeap(int capacity, int* numbers)
{
    Heap* h = (Heap*)malloc(sizeof(Heap));
    if (h == NULL) {
        printf("Memory allocation error");
        return NULL;
    }

    h->size = 0;
    h->capacity = capacity;

    // Allocate memory for the array (+1 because index 0 will not be used)
    h->array = (int*)malloc((capacity + 1) * sizeof(int));
    if (h->array == NULL) {
        printf("Memory allocation error");
        return NULL;
    }

    // Copy elements into the array starting at index 1
    for (int i = 0; i < capacity; i++) {
        h->array[i + 1] = numbers[i];
    }

    h->size = capacity;

    // Build the Heap: run Heapify from the last parent down to the root
    int i = h->size / 2;
    while (i >= 1) {
        maxHeapify(h, i);
        i--;
    }

    /*
      Visualization of how the Heap is stored in the array:

               array[1]
             /           \
        array[2]       array[3]
       /      \        /      \
    array[4] array[5] array[6] array[7]
     /   \
  array[8] array[9]

      Note: array[0] is unused.
    */

    return h;
}
```

```c++
// Helper function to maintain Heap property when inserting
// Moves a node up while it's larger than its parent
void bubbleUp(Heap* h, int index)
{
    int parent = index / 2;
    if (parent >= 1 && h->array[parent] < h->array[index]) {
        // Swap child and parent if child is greater
        int temp = h->array[parent];
        h->array[parent] = h->array[index];
        h->array[index] = temp;

        // Continue checking the parent recursively
        bubbleUp(h, parent);
    }
}
```

```c++
// Function to remove and return the maximum element
int extractMax(Heap* h)
{
    if (h->size == 0) {
        printf("\nHeap is empty.");
        return -999;
    }

    int maxElement = h->array[1];  // Root element (maximum in max-Heap)

    // Move the last element to the root
    h->array[1] = h->array[h->size];
    h->size--;

    // Restore Heap property by Heapifying down from root
    maxHeapify(h, 1);

    /*
      After extracting:
      - The last element moves to the root
      - Heap is restructured with maxHeapify
      - Index 0 remains unused
    */

    return maxElement;
}
```

```c++
// Function to insert a new value into the Heap
void insert(Heap* h, int value)
{
    if (h->size < h->capacity) {
        h->size++;
        h->array[h->size] = value;
        bubbleUp(h, h->size);  // Move the new value up if needed
    }
}
```

```c++
// Function to print the Heap contents
void printHeap(Heap* h)
{
    printf("Heap contents (indices 1 to %d):\n", h->size);
    for (int i = 1; i <= h->size; i++) {
        printf("%d ", h->array[i]);
    }
    printf("\n");
}
```

```c++
int inputArray[9] = {1,2,3,4,5,6,7,8,9};

Heap* h = createHeap(9, inputArray);

printf("After building the Heap:\n");
printHeap(h);

printf("\nExtracting max (%d):\n", extractMax(h));
printHeap(h);

printf("\nInserting 15:\n");
insert(h, 15);
printHeap(h);

printf("\nExtracting new max (%d):\n", extractMax(h));
printHeap(h);
```
