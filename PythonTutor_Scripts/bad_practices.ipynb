## Morris Worm (1988)

Libraries responsible for e-mail management had code that looked similar to excerpt below.

It took out around 6000 machines (10% of the internet by that time)

We are not checking if the input fits inside the buffer, this could allow a Buffer Overflow (similar concept to Pokemon glitch but not quite the same)

It can go on to overwrite instructions out of bounds, allowing for Remote Code Execution. C does not perform automatic bounds checking

```python
void vulnerable_func( char *input){
    char buffer[8]; // Buffer with 8 bytes

    strcpy(buffer, input);
    //Fix: strcpy(buffer, input, sixeof(buffer-1));
    printf("You entered: %s\n", buffer);
}

int main(int argc, char *argv[]){

    vulnerable_func(argv[1]);

    return 0;
}
```
