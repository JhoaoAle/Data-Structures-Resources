## Handling Dynamic Memory Allocations in C

Dynamic memory allocation allows a programmer to allocate, resize, and free memory at runtime.

### malloc

The malloc() (stands for **m**emory **alloc**ation) function is used to allocate a single block of contiguous memory on the heap at runtime. The memory allocated by malloc() is uninitialized, meaning it contains garbage values.

Assume that we want to create an array to store 5 integers. Since the size of int is 4 bytes, we need $5 * 4 bytes = 20$ bytes of memory

```c++
int array_size = 5;
int *ptr = (int *)malloc(sizeof(int) * array_size);

if (ptr == NULL) {
        printf("Allocation Failed");
        exit(0);
    }

printf("Printing array before assigning values: ");
for (int i = 0; i < array_size; i++) printf("\n %d ", ptr[i]);
```

```c++
for (int i = 0; i < array_size; i++) ptr[i] = i + 1;

printf("Printing array after assigning values: ");
for (int i = 0; i < array_size; i++) printf("\n %d ", ptr[i]);
```

```c++
memset(ptr, 0, array_size * sizeof(int));
printf("Printing array after using memset: ");
for (int i = 0; i < array_size; i++) printf("\n %d ", ptr[i]);
```

```c++
// Freeeing memory after usage
free(ptr);
```

### calloc

The calloc() (stands for **c**ontiguous **alloc**ation) function is similar to malloc(), but it initializes the allocated memory to zero. It is used when you need memory with default zero values. It essentially works as running memset after malloc.

```c++
int *ptr_calloc = (int *)calloc(array_size, sizeof(int));

if (ptr_calloc == NULL) {
    printf("Allocation Failed");
    exit(0);
}
    
printf("Printing array before assigning values: ");
for (int i = 0; i < array_size; i++) printf("\n %d ", ptr_calloc[i]);


for (int i = 0; i < array_size; i++) ptr_calloc[i] = i + 1;
printf("\n Printing array after assigning values: ");
for (int i = 0; i < array_size; i++) printf("\n %d ", ptr_calloc[i]);
```

```c++
free(ptr_calloc);
```

### realloc

The realloc() function is used to **re**size a previously **alloc**ated memory block. It allows to change the size of an existing memory allocation without needing to free the old memory and allocate a new block.

Suppose it was initially allocated a total of 5 integers space in memory, but later it is needed to expand the array to hold 10 integers. realloc() can be used to resize the memory block:

```c++
int *ptr_realloc = (int *)malloc(5 * sizeof(int));
    
// Check if allocation was successful
if (ptr_realloc == NULL) {
    printf("Memory Allocation Failed\n");
    exit(0);
}

// Now, we need to store 8 elements so
// Reallocate to store 8 integers
ptr_realloc = (int *)realloc(ptr_realloc, 8 * sizeof(int)); 

// Check if reallocation was successful
if (ptr_realloc == NULL) {
    printf("Memory Reallocation Failed\n");
    exit(0);
}

// Assume we only use 5 elements now
for (int i = 0; i < 5; i++) ptr_realloc[i] = (i + 1) * 10;


// Shrink the array back to 5 elements
ptr_realloc = (int *)realloc(ptr_realloc, 5 * sizeof(int));

// Check if shrinking was successful
if (ptr_realloc == NULL) {
    printf("Memory Reallocation Failed\n");
    exit(0);
}

for (int i = 0; i < 5; i++) printf("%d ", ptr_realloc[i]);

// Finally, free the memory when done
free(ptr_realloc);
```

## Potential Risks with irresponsible Memory management

- **Memory Leaks:** Failing to free dynamically allocated memory leads to memory leaks, exhausting system resources.
- **Dangling Pointers:** Using a pointer after freeing its memory can cause undefined behavior or crashes.
- **Fragmentation:** Repeated allocations and deallocations can fragment memory, causing inefficient use of heap space.

```c++
printf("Step 1: Allocating memory for 100 integers\n");
int *numbers = (int *)malloc(100 * sizeof(int));

if (numbers == NULL) {
    printf("Memory allocation failed\n");
    exit(0);
}

for (int i = 0; i < 100; i++) numbers[i] = i * 5;

printf("First allocation: numbers[0]=%d, numbers[99]=%d\n", numbers[0], numbers[99]);


int *lost = numbers;

printf("\nStep 2: Allocating memory again WITHOUT freeing first allocation\n");
numbers = (int *)malloc(200 * sizeof(int));

if (numbers == NULL) {
    printf("Memory allocation failed\n");
    exit(0);
}

// Initialize the new array
for (int i = 0; i < 200; i++) numbers[i] = i * 2;

printf("Second allocation: numbers[0]=%d, numbers[199]=%d\n", numbers[0], numbers[199]);

// Attempting to access the old memory
printf("lost[0]=%d, lost[99]=%d\n", lost[0], lost[99]);

free(numbers);
```
