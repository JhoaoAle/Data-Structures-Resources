## What we already know

```c++
typedef struct Node {
    int data;
    struct Node* next;
} Node;
```

```c++
Node* create_node(int value){
    Node* new_Node = (Node *)malloc(sizeof(Node));
    new_Node->data = value;
    new_Node->next = NULL;
    return new_Node;
}
```

## Introducing Stacks

A stack is an abstract data type that serves as a collection of elements with two main operations:

- Push, which adds an element to the collection, and
- Pop, which removes the most recently added element.

Additionally, a peek operation can, without modifying the stack, return the value of the last element added (the item at the top of the stack). The name stack is an analogy to a set of physical items stacked one atop another, such as a stack of plates.

The order in which an element added to or removed from a stack is described as **Last-In-First-Out (LIFO)**. As with a stack of physical objects, this structure makes it easy to take an item off the top of the stack, but accessing a datum deeper in the stack may require removing multiple other items first.

Considered a sequential collection, a stack has one end which is the only position at which the push and pop operations may occur, the top of the stack, and is fixed at the other end, the bottom.

A stack may be implemented to have a bounded capacity. 

If the stack is full and does not contain enough space to accept another element, the stack is in a state of **stack overflow**. 

```c++
typedef struct Stack {
    Node* top;
} Stack;
```

```c++
Stack* create_stack(){
    Stack* stack = (Stack *)malloc(sizeof(Stack));
    stack->top = NULL;
    return stack;
}
```

```c++
bool is_empty_stack(Stack* stack){
    return (stack->top == NULL);
}
```

## New Stack Operations

### Push

```c++
void push(Stack* stack, int value){
    Node* new_node = create_node(value);

    new_node->next = stack->top;
    stack->top = new_node;
}
```

### Pop

```c++
int pop(Stack* stack){
    if(stack->top == NULL){
        printf("Stack is empty\n");
        return -1;
    }

    Node* temp = stack->top;
    int value = temp->data;

    stack->top = stack->top->next;
    free(temp);

    return value;
}
```

### Peek

```c++
int peek(Stack* stack){
    if(stack->top == NULL){
        printf("Stack is empty\n");
        return -1;
    }

    return stack->top->data;
}
```

## Printing and Memory Cleanup

```c++
void print_stack(Stack* stack){
    if(stack->top == NULL){
        printf("NULL\n");
        return;
    }

    Node* temp = stack->top;

    while(temp != NULL){
        printf("%d -> ", temp->data);
        temp = temp->next;
    }

    printf("NULL\n");
}
```

```c++
void free_stack(Stack* stack){
    Node* current = stack->top;
    Node* temp;

    while(current != NULL){
        temp = current;
        current = current->next;
        free(temp);
    }

    free(stack);
}
```

## Testing

```c++
Stack* my_stack = create_stack();

push(my_stack, 10);
push(my_stack, 20);
push(my_stack, 30);

print_stack(my_stack);
```

```c++
printf("Popped: %d\n", pop(my_stack));
print_stack(my_stack);
```

```c++
printf("Top: %d\n", peek(my_stack));
```

```c++
free_stack(my_stack);
```
