#### 3.

Implementar un código el cual determine si una expresión que contiene paréntesis (), corchetes [] y/o llaves {} se encuentra bien balanceada; esto es, que todo símbolo de apertura tenga posteriormente un símbolo de cierre en el orden correcto

Ejemplos:

$\{ [ ( ) ] \}$

$\{ ( [ ) ] \}$

$[ ( a + b ) * \{ c / d \} ]$

Para el primer y tercer casos debe retornar 'true' y para el segundo caso debe retornar 'false'.

##### Array based

```c++
typedef struct {
    char data[MAX];
    int top;
} StackArray;
```

```c++
void initialize_stack(StackArray *p) {
    p->top = -1;
}
```

```c++
bool is_empty(StackArray *p) {
    return p->top == -1;
}
```

```c++
void push(StackArray *p, char c) {
    if (p->top < MAX - 1) {
        p->data[++(p->top)] = c;
    }
}
```

```c++
char pop(StackArray *p) {
    if (!is_empty(p)) {
        return p->data[(p->top)--];
    }
    return '\0';
}
```

```c++
bool is_pair(char open, char close) {
    return (open == '(' && close == ')') ||
           (open == '[' && close == ']') ||
           (open == '{' && close == '}');
}
```

```c++
bool is_balanced_array_based(char *expression) {
    StackArray p;
    initialize_stack(&p);

    for (int i = 0; i < strlen(expression); i++) {
        char c = expression[i];

        if (c == '(' || c == '[' || c == '{') {
            push(&p, c);
        } else if (c == ')' || c == ']' || c == '}') {
            if (is_empty(&p)) return false;
            char top = pop(&p);
            if (!is_pair(top, c)) return false;
        }
    }

    return is_empty(&p);
}
```

```c++
char exp1[] = "{[()]}";
char exp2[] = "{([)]}";
char exp3[] = "[(a+b)*{c/d}]";

printf("%s -> %s\n", exp1, is_balanced_array_based(exp1) ? "true" : "false");
printf("%s -> %s\n", exp2, is_balanced_array_based(exp2) ? "true" : "false");
printf("%s -> %s\n", exp3, is_balanced_array_based(exp3) ? "true" : "false");
```

##### Linked List based

```c++
typedef struct Node {
    char data;
    struct Node *next;
} Node;
```

```c++
typedef struct {
    Node *top;
} StackList;
```

```c++
void initialize_stack_node(StackList *p) {
    p->top = NULL;
}
```

```c++
bool is_empty_s(StackList *p) {
    return p->top == NULL;
}
```

```c++
void push_s(StackList *p, char value) {
    Node *new_node = (Node *)malloc(sizeof(Node));
    new_node->data = value;
    new_node->next = p->top;
    p->top = new_node;
}
```

```c++
char pop_s(StackList *p) {
    if (is_empty(p)) return '\0';

    Node *temp = p->top;
    char value = temp->data;
    p->top = temp->next;
    free(temp);

    return value;
}
```

```c++
bool is_balanced_stack_based(char *expression) {
    StackList p;
    initialize_stack_node(&p);

    for (int i = 0; expression[i] != '\0'; i++) {
        char c = expression[i];

        if (c == '(' || c == '[' || c == '{') {
            push_s(&p, c);
        } else if (c == ')' || c == ']' || c == '}') {
            if (is_empty_s(&p)) return false;

            char top = pop_s(&p);
            if (!is_pair(top, c)) return false;
        }
    }

    return is_empty_s(&p);
}
```

```c++
char exp1[] = "{[()]}";
char exp2[] = "{([)]}";
char exp3[] = "[(a+b)*{c/d}]";

printf("%s -> %s\n", exp1, is_balanced_stack_based(exp1) ? "true" : "false");
printf("%s -> %s\n", exp2, is_balanced_stack_based(exp2) ? "true" : "false");
printf("%s -> %s\n", exp3, is_balanced_stack_based(exp3) ? "true" : "false");
```
