**1. (1.0)** En el siguiente ejemplo se propone la implementación de un algoritmo de ordenamiento que ordene los elementos de un array de manera **ascendente**:

```
Procedimiento BubbleSort(arr, n)
    Repetir
        swapped ← falso
        Para i desde 0 hasta n - 2 hacer
           Si arr[i] < arr[i+1] entonces
                temp ← arr[i]
                arr[i] ← arr[i+1]
                arr[i+1] ← temp
                swapped ← verdadero
            FinSi
        FinPara
    Hasta que swapped = falso
FinProcedimiento
```

````c++
void BubbleSort(int arr[], int n) {
    bool swapped;
    do {
        swapped = false;

        for (int i = 0; i <= n - 2; i++) {
            if (arr[i] < arr[i + 1]) {
                int temp = arr[i];
                arr[i] = arr[i + 1];
                arr[i + 1] = temp;
                swapped = true;
            }
        }
    } while (swapped == true);
}
````

````c++
int arr[] = {5, 1, 4, 8, 2};
int n = 5;

BubbleSort(arr, n);

for (int i = 0; i < n; i++) printf("%d ", arr[i]);
````

### Primer enfoque: Arrays

````c++
bool palindrome_array(char str[]) {
    int inicio = 0;
    int fin = strlen(str) - 1;

    while (inicio < fin) {
        if (str[inicio] != str[fin]) {
            return false;
        }
        inicio++;
        fin--;
    }
    return true;
}
````

### Segundo enfoque: Lista Doblemente Enlazada

````c++
typedef struct DLL_Node {
    char data;
    struct DLL_Node* prev;
    struct DLL_Node* next;
} DLL_Node;
````

````c++
DLL_Node* create_DLL_Node(char c) {
    DLL_Node* new_node = (DLL_Node*)malloc(sizeof(DLL_Node));
    new_node->data = c;
    new_node->prev = NULL;
    new_node->next = NULL;
    return new_node;
}
````

````c++
bool palindrome_DLL(char str[]) {
    DLL_Node *start = NULL, *end = NULL;

    for(int i = 0; i < strlen(str); i++) {
        DLL_Node* new_node = create_DLL_Node(str[i]);

        if(start == NULL) start = end = new_node;
        else {
            end->next = new_node;
            new_node->prev = end;
            end = new_node;
        }
    }

    DLL_Node* left = start;
    DLL_Node* right = end;

    while(left != right && left->prev != right) {
        if(left->data != right->data)
            return false;

        left = left->next;
        right = right->prev;
    }
    return true;
}
````

### Tercer enfoque: Pilas

````c++
typedef struct Node {
    char data;
    struct Node* next;
} Node;

typedef struct Stack {
    Node* top;
} Stack;
````

````c++
void push(Stack* s, char c) {
    Node* new_node = (Node*)malloc(sizeof(Node));
    new_node->data = c;
    new_node->next = s->top;
    s->top = new_node;
}
````

````c++
char pop(Stack* s) {
    if (s->top == NULL)
        return '\0';

    Node* temp = s->top;
    char value = temp->data;

    s->top = temp->next;
    free(temp);

    return value;
}
````

````c++
bool palindrome_stack(char str[]) {
    Stack* s;
    s->top = NULL;

    int n = strlen(str);
    for(int i = 0; i < n; i++) {
        push(s, str[i]);
    }

    for(int i = 0; i < n; i++) {
        if(str[i] != pop(s))
            return false;
    }
    return true;
}
````

### Cuarto enfoque: Lista Enlazada

````c++
typedef struct LinkedList{
    Node* head;
    Node* tail;
} LinkedList;
````

````c++
void insert_head(LinkedList* list, char c) {
    Node* new_node = (Node*)malloc(sizeof(Node));
    new_node->data = c;
    new_node->next = list->head;

    list->head = new_node;

    if (list->tail == NULL) {
        list->tail = new_node;
    }
}
````

````c++
void insert_tail(LinkedList* list, char c) {
    Node* new_node = (Node*)malloc(sizeof(Node));
    new_node->data = c;
    new_node->next = NULL;

    if (list->tail == NULL) {
        list->head = list->tail = new_node;
        return;
    }

    list->tail->next = new_node;
    list->tail = new_node;
}
````

````c++
bool palindrome_LL(char str[]) {
    LinkedList* list_head = (LinkedList*)malloc(sizeof(LinkedList));
    LinkedList* list_tail = (LinkedList*)malloc(sizeof(LinkedList));

    list_head->head = list_head->tail = NULL;
    list_tail->head = list_tail->tail = NULL;

    int n = strlen(str);

    for(int i = 0; i < n; i++) {
        insert_head(list_head, str[i]);
        insert_tail(list_tail, str[i]);
    }

    Node* p1 = list_head->head;
    Node* p2 = list_tail->head;

    while(p1 != NULL && p2 != NULL) {
        if(p1->data != p2->data)
            return false;
        p1 = p1->next;
        p2 = p2->next;
    }

    return true;
}
````
